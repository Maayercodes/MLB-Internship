```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('CleanedStudentPerformance.csv')
print(df.head())
```

```python
plt.figure(figsize=(12, 6))

plt.bar(df.index, df["Average_Score"])

plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.title("Average Marks of Each Student")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Average_Score"], bins=10)

plt.xlabel("Average Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Average Scores")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["Math_Score"], df["Reading_Score"])

plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.title("Math vs Reading Scores")

plt.show()
```

```python
performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    performance_counts,
    labels=performance_counts.index,
    autopct="%1.1f%%"
)

plt.title("Students by Performance Category")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.boxplot([
    df["Math_Score"],
    df["Reading_Score"],
    df["Writing_Score"]
], labels=["Math", "Reading", "Writing"])

plt.ylabel("Marks")
plt.title("Distribution of Marks in All Subjects")

plt.show()
```
