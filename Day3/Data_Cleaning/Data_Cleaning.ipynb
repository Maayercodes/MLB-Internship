```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('StudentsPerformance.csv')
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df["math score"] = df["math score"].fillna(df["math score"].mean())
```

```python
print(df["math score"].isnull().sum())
```

```python
df["reading score"] = df["reading score"].fillna(df["reading score"].mean())
```

```python
print(df["reading score"].isnull().sum())
```

```python
df["writing score"] = df["writing score"].fillna(df["writing score"].mean())
```

```python
print(df["writing score"].isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
df = df.rename(columns={
    "race/ethnicity": "Race_Ethnicity",
    "parental level of education": "Parental_Education",
    "test preparation course": "Test_Preparation",
    "math score": "Math_Score",
    "reading score": "Reading_Score",
    "writing score": "Writing_Score"
})

print(df.columns)
```

```python
df["Average_Score"] = df[["Math_Score", "Reading_Score", "Writing_Score"]].mean(axis=1)

print(df[["Math_Score", "Reading_Score", "Writing_Score", "Average_Score"]].head())
```

```python
conditions = [
    df["Average_Score"] >= 90,
    df["Average_Score"] >= 80,
    df["Average_Score"] >= 70,
    df["Average_Score"] < 70
]

choices = [
    "Excellent",
    "Good",
    "Average",
    "Needs Improvement"
]

df["Performance"] = np.select(conditions, choices, default="Needs Improvement")

print(df[["Average_Score", "Performance"]].head(10))
```

```python
df.to_csv("CleanedStudentPerformance.csv", index=False)

print("Cleaned dataset saved successfully!")
```
